```python
# Define the path to the ZIP file and the directory to extract the images to
zip_file_path = "HAM10000_allimage.zip"
extract_dir_path = os.path.join(os.getcwd(), "HAM10000_images")

# Create the directory to extract the images to, if it doesn't exist already
if not os.path.exists(extract_dir_path):
    os.makedirs(extract_dir_path)

# Open the ZIP file
with ZipFile(zip_file_path, 'r') as zip_file:
    # Loop through all the files in the ZIP archive
    for filename in zip_file.namelist():
        # Check if the file is a JPEG image
        if filename.lower().endswith('.jpg'):
            # Extract the file to the target directory
            zip_file.extract(filename, extract_dir_path)
```

# CNN

```python
# Creating a CNN model
model = tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(36, (3, 3), activation='relu', input_shape=(64, 64, 3)),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Conv2D(72, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Conv2D(144, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(288, activation='relu'),
    tf.keras.layers.Dense(288, activation='relu'),
    tf.keras.layers.Dense(7, activation='softmax')
])
```

```python
# Compiling model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
# Training model
trained_model = model.fit(x_train, y_train, epochs=45, validation_data=(x_test, y_test))
```

```python
# Evaluating model on test set
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print("Test accuracy:", test_acc)
```

```python
# graphical representaion of training and validation loss
loss = trained_model.history['loss']
val_loss = trained_model.history['val_loss']

epochs = range(len(loss))

plt.plot(epochs, loss, 'r', label='Training loss')
plt.plot(epochs, val_loss, 'b', label='Validation loss')
plt.title('Training and validation loss')
plt.legend(loc=0)
plt.figure()


plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras.applications.resnet import ResNet50

# Load pre-trained ResNet50 model
model = ResNet50(weights='imagenet', include_top=False, input_shape=(64, 64, 3))

# Add classification layers on top of ResNet50
x = tf.keras.layers.GlobalAveragePooling2D()(model.output)
x = tf.keras.layers.Dense(128, activation='relu')(x)
x = tf.keras.layers.Dropout(0.5)(x)
output = tf.keras.layers.Dense(1, activation='softmax')(x)

# Define the new model with classification layers added
model = tf.keras.models.Model(inputs=model.input, outputs=output)

# Compile the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```
